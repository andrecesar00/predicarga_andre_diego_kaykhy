```r
# MÓDULO 9 — CONGELAR O CORPUS EM 64 DOCUMENTOS

# Fonte declarada na ficha do grupo
fonte <- "Wikipedia pt"

# Monta a tabela no formato exigido pela ferramenta julgar.html
bruto <- data.frame(
  id     = names(docs),      # d1, d2, ..., d64
  titulo = origem,           # artigo de origem de cada parágrafo
  texto  = docs,             # o parágrafo
  fonte  = fonte,            # mesma fonte para todas as linhas
  stringsAsFactors = FALSE   # mantém texto como texto
)

# Cria a URL da Wikipédia a partir do título do artigo
bruto$url <- paste0(
  "https://pt.wikipedia.org/wiki/",
  gsub(" ", "_", origem)
)

# Normaliza sem gerar ids novos: mantém d1, d2, ..., d64
cp <- cc$normalizar(
  bruto,
  col_texto  = "texto",
  col_titulo = "titulo",
  col_id     = "id"
)

# Ficha técnica do corpus
cc$validar(cp)

# Confirmação dos ids originais
cp$id[1:3]

# Quantidade final de documentos
nrow(cp)
```
